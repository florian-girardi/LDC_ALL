```python
# List to store pivoted DataFrames
pivoted_list = []

# Process Panamax
for df, name in zip(dfs_pmx, df_names_pmx):
    parts = name.split('_')
    origin = parts[1].upper()
    destination = parts[2].upper()

    df['origin'] = origin
    df['destination'] = destination
    df['month_year'] = df['contract_month'].astype(str) + '-' + df['contract_year'].astype(str)
    df['vessel_size'] = "PANAMAX"

    pivot_df = df.pivot_table(
        index=['destination','origin','vessel_size'],
        columns='month_year',
        values='value',
        aggfunc='first'
    ).reset_index()

    # Sort month-year columns chronologically
    month_cols = [c for c in pivot_df.columns if c not in ['destination','origin','vessel_size']]
    month_cols_sorted = sorted(
        month_cols,
        key=lambda x: pd.to_datetime(x, format='%m-%Y')
    )

    pivot_df = pivot_df[['destination','origin','vessel_size'] + month_cols_sorted]
    pivoted_list.append(pivot_df)

# Process Supramax
for df, name in zip(dfs_smx, df_names_smx):
    parts = name.split('_')
    origin = parts[1].upper()
    destination = parts[2].upper()

    df['origin'] = origin
    df['destination'] = destination
    df['month_year'] = df['contract_month'].astype(str) + '-' + df['contract_year'].astype(str)
    df['vessel_size'] = "SUPRAMAX"

    pivot_df = df.pivot_table(
        index=['destination','origin','vessel_size'],
        columns='month_year',
        values='value',
        aggfunc='first'
    ).reset_index()

    # Sort month-year columns chronologically
    month_cols = [c for c in pivot_df.columns if c not in ['destination','origin','vessel_size']]
    month_cols_sorted = sorted(
        month_cols,
        key=lambda x: pd.to_datetime(x, format='%m-%Y')
    )

    pivot_df = pivot_df[['vessel_size','destination','origin'] + month_cols_sorted]
    pivoted_list.append(pivot_df)

# Concatenate all into a single DataFrame
final_df = pd.concat(pivoted_list, ignore_index=True)
```

```python
# Mapping dictionary
destination_mapping = {
    'DA': 'Dalian',
    'QI': 'Qingdao',
}

origin_mapping = {
    'UPBB': 'UPBB',
    'MONTE': 'MONTEVIDEO',
    'PNW': 'PNW (SEATTLE)',
    'MONTENUEVA': 'MONTEVIDEO + NUEVA PALMIRA',
    'SAN': 'SANTOS',
    'USG': 'USG',
    'MONTE': 'MONTEVIDEO',
    
}
# Apply mapping to the destination column
final_df['destination'] = final_df['destination'].map(destination_mapping)
final_df['origin'] = final_df['origin'].map(origin_mapping)
final_df = final_df.sort_values(by=['vessel_size','destination','origin'])
final_df.reset_index(drop=True)

excel_buffer_final_df = BytesIO()
final_df.to_excel(excel_buffer_final_df, index=False, sheet_name='Rates')
excel_buffer_final_df.seek(0)  # Reset the buffer's position
```

```python
# Keep shifting start_date back until dataframe is not empty
while True:

    upbb_qi_pmx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-PANAMAX-SOYBEANS-AR, Upriver+Bahia Blanca-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    upbb_qi_smx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-SUPRAMAX-SOYBEANS-AR, Upriver+Bahia Blanca-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")

    monte_qi_pmx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-PANAMAX-SOYBEANS-UY, Montevideo-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    monte_qi_smx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-SUPRAMAX-SOYBEANS-UY, Montevideo-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")

    montenueva_qi_pmx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-PANAMAX-SOYBEANS-UY, Nueva Palmira+Montevideo-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")


    usg_qi_pmx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-PANAMAX-SOYBEANS-US, Port Allen-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    usg_qi_smx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-SUPRAMAX-SOYBEANS-US, Port Allen-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    
    pnw_qi_pmx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-PANAMAX-SOYBEANS-US, Seattle-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    pnw_qi_smx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-SUPRAMAX-SOYBEANS-US, Seattle-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    
    san_qi_pmx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-PANAMAX-SOYBEANS-BR, Santos-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    san_qi_smx=zema.get_curve(curve='P-FREIGHT-LDC-INPUT-SUPRAMAX-SOYBEANS-BR, Santos-CN, Qingdao-USD-MT', period=f"{start_date}::{end_date}")
    
    if not upbb_qi_pmx.empty:
        break
    else:
        start_date -= timedelta(days=1)

dfs = [
    upbb_qi_pmx, upbb_qi_smx,
    monte_qi_pmx, monte_qi_smx,
    montenueva_qi_pmx,
    usg_qi_pmx, usg_qi_smx,
    pnw_qi_pmx, pnw_qi_smx,
    san_qi_pmx, san_qi_smx
]

df_names = [
    "upbb_qi_pmx", "upbb_qi_smx",
    "monte_qi_pmx", "monte_qi_smx",
    "montenueva_qi_pmx",
    "usg_qi_pmx", "usg_qi_smx",
    "pnw_qi_pmx", "pnw_qi_smx",
    "san_qi_pmx", "san_qi_smx"
]


# === Step 2: Function to process each DataFrame ===
def process_df_hist(df, name):
    df = df[df["observation"] == "Last"]
    df['month_year'] = df['contract_month'].astype(str) + '-' + df['contract_year'].astype(str)
    df = df[["date", "value", "month_year"]]
    
    # Add Vessel column based on name
    if "pmx" in name.lower():
        df["Vessel"] = "PANAMAX"
    elif "smx" in name.lower():
        df["Vessel"] = "SUPRAMAX"
    else:
        df["Vessel"] = None  # fallback if neither is found
    
    return df

# Build dictionary of processed DataFrames
hist_dfs = {}
for df, name in zip(dfs, df_names):
    hist_dfs[f"hist_{name}"] = process_df_hist(df, name)

# Argentina - UPBB
hist_upbb_qi_pmx = hist_dfs["hist_upbb_qi_pmx"]
hist_upbb_qi_smx = hist_dfs["hist_upbb_qi_smx"]

# Uruguay - Montevideo
hist_monte_qi_pmx = hist_dfs["hist_monte_qi_pmx"]
hist_monte_qi_smx = hist_dfs["hist_monte_qi_smx"]

# Uruguay - Nueva Palmira + Montevideo
hist_montenueva_qi_pmx = hist_dfs["hist_montenueva_qi_pmx"]

# # US Gulf - Port Allen
hist_usg_qi_pmx = hist_dfs["hist_usg_qi_pmx"]
hist_usg_qi_smx = hist_dfs["hist_usg_qi_smx"]

# PNW - seattle
hist_pnw_qi_pmx = hist_dfs["hist_pnw_qi_pmx"]
hist_pnw_qi_smx = hist_dfs["hist_pnw_qi_smx"]

# Santos - Brazil
hist_san_qi_pmx = hist_dfs["hist_san_qi_pmx"]
hist_san_qi_smx = hist_dfs["hist_san_qi_smx"]

hist_upbb_qi=pd.concat([hist_upbb_qi_pmx,hist_upbb_qi_smx])
hist_upbb_qi['Origin']='UPBB'

hist_monte_qi=pd.concat([hist_monte_qi_pmx,hist_monte_qi_smx])
hist_monte_qi['Origin']='MONTEVIDEO'

hist_montenueva_qi=hist_montenueva_qi_pmx
hist_montenueva_qi['Origin']='MONTEVIDEO + NUEVA PALMIRA'


hist_usg_qi=pd.concat([hist_usg_qi_pmx,hist_usg_qi_smx])
hist_usg_qi['Origin']='USG'

hist_pnw_qi=pd.concat([hist_pnw_qi_pmx,hist_pnw_qi_smx])
hist_pnw_qi['Origin']='PNW (SEATTLE)'

hist_san_qi=pd.concat([hist_san_qi_pmx,hist_san_qi_smx])
hist_san_qi['Origin']='SANTOS'
hist_df=pd.concat([hist_upbb_qi,hist_monte_qi,hist_montenueva_qi,hist_usg_qi,hist_pnw_qi,hist_san_qi])


pivot_df_hist = hist_df.pivot_table(
    index=['date','Origin','Vessel'],
    columns='month_year',
    values='value',
    aggfunc='first'
).reset_index()
pivot_df_hist['Destination']='QINGDAO'

pivot_df_hist = pivot_df_hist.sort_values(by=['date','Vessel','Origin'])
pivot_df_hist.reset_index(drop=True)

fixed_cols = ["date","Destination", "Vessel","Origin"]
contract_cols = [c for c in pivot_df_hist.columns if c not in fixed_cols]

# Convert "M-YYYY" to datetime for sorting
def parse_contract(c):
    try:
        month, year = c.split("-")
        return pd.to_datetime(f"{year}-{month}-01")
    except:
        return pd.NaT

sorted_contracts = sorted(contract_cols, key=parse_contract)

# Reorder DataFrame
pivot_df_hist = pivot_df_hist[fixed_cols + sorted_contracts]

pivot_df_hist.reset_index(drop=True)

excel_buffer_pivot_df_hist = BytesIO()
pivot_df_hist.to_excel(excel_buffer_pivot_df_hist, index=False, sheet_name='Rates')
excel_buffer_pivot_df_hist.seek(0)  # Reset the buffer's position

```

```python

df_long = pivot_df_hist.melt(
    id_vars=["date", "Destination", "Vessel", "Origin"],
    var_name="contract",
    value_name="freight_rate"
)

# Convert date
df_long["date"] = pd.to_datetime(df_long["date"], errors="coerce")

# Parse contracts into datetime for sorting
df_long["contract_date"] = pd.to_datetime(df_long["contract"], format="%m-%Y", errors="coerce")

# Keep valid values
df_long = df_long.dropna(subset=["freight_rate", "contract_date"])

# Color palette for contracts
palette = px.colors.qualitative.Set2 + px.colors.qualitative.Set1 + px.colors.qualitative.Pastel

# Loop over origins
figs = {}
for origin in df_long["Origin"].unique():
    fig = go.Figure()

    contracts_sorted = sorted(df_long["contract"].unique(),
                              key=lambda x: pd.to_datetime(x, format="%m-%Y"))
    color_map = {c: palette[i % len(palette)] for i, c in enumerate(contracts_sorted)}

    for vessel in ["PANAMAX", "SUPRAMAX"]:
        df_sub = df_long[(df_long["Origin"] == origin) & (df_long["Vessel"] == vessel)]
        if df_sub.empty:
            continue
        for contract in contracts_sorted:
            df_c = df_sub[df_sub["contract"] == contract]
            if df_c.empty:
                continue
            fig.add_trace(go.Scatter(
                x=df_c["date"],
                y=df_c["freight_rate"],
                mode="lines",
                name=f"{vessel} - {contract}",
                line=dict(
                    color=color_map[contract],
                    dash="solid" if vessel == "PANAMAX" else "dot"
                )
            ))

    # Buttons to toggle Panamax/Supramax
    buttons = []
    for vessel in ["PANAMAX", "SUPRAMAX", "ALL"]:
        visible = []
        for trace in fig.data:
            if vessel == "ALL":
                visible.append(True)
            else:
                visible.append(vessel in trace.name)
        buttons.append(dict(
            label=vessel,
            method="update",
            args=[{"visible": visible}]
        ))
        # Additional buttons for each contract
    contracts_sorted = sorted(df_long["contract"].unique(),
                            key=lambda x: pd.to_datetime(x, format="%m-%Y"))

    for contract in contracts_sorted:
        visible = []
        for trace in fig.data:
            # show only traces containing this contract (both PANAMAX and SUPRAMAX)
            if contract in trace.name:
                visible.append(True)
            else:
                visible.append(False)
        buttons.append(dict(
            label=contract,
            method="update",
            args=[{"visible": visible}]
        ))
        
    fig.update_layout(
        title=f"Freight Rate Evolution - {origin}",
        xaxis_title="Date",
        yaxis_title="Freight Rate",
        hovermode="x unified",
        updatemenus=[dict(
        type="buttons",
        direction="left",       # buttons laid out horizontally
        x=0,                    # x=0 starts at the left of the plot
        y=1.15,                 # y>1 to place above the plot
        xanchor="left",
        yanchor="top",
        buttons=buttons,
        showactive=True
        )]
    )

    figs[origin] = fig

# Display each chart
for origin, fig in figs.items():
    fig.show()

```
