#### AVERAGE OVER EARLY STAGES

```python
def assign_marketing_year(row):
    year = row.date.year
    month = row.date.month
    if month < 12:
        return f"{year-1} / {year}"
    else:
        return f"{year} / {year+1}"

arg['date'] = pd.to_datetime(arg['date'])
arg['Marketing Year'] = arg.apply(assign_marketing_year, axis=1)


# Extract month
arg['month'] = arg['date'].dt.month
arg['year'] = arg['date'].dt.year

arg_filt=arg[arg['year'] >= 2010]
# Keep only May to August
early_stage_df = arg_filt[arg_filt['month'].between(5, 8)]

early_stage_avg = early_stage_df.groupby("Marketing Year")['value'].mean().reset_index()
early_stage_avg.columns = ['Marketing Year', 'Early Stage Soil Moisture']

# Example yield_df
# yield_df = pd.DataFrame({"Marketing Year": [...], "Yield": [...]})
merged_df = pd.merge(yield_arg, early_stage_avg, on="Marketing Year", how="inner")

correlation = merged_df['Yield'].corr(merged_df['Early Stage Soil Moisture'])
print(f"Correlation (Yield vs Early Soil Moisture): {correlation:.3f}")

# Optional: visualize
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.regplot(data=merged_df, x='Early Stage Soil Moisture', y='Yield')

# Add labels for each point
for i, row in merged_df.iterrows():
    plt.text(
        row['Early Stage Soil Moisture'], 
        row['Yield'], 
        row['Marketing Year'], 
        fontsize=8,
        ha='right',  # horizontal alignment
        va='bottom'  # vertical alignment
    )

plt.title('Wheat Yield vs Early-Stage Soil Moisture (May–Aug)')
plt.xlabel('Soil Moisture (May–Aug)')
plt.ylabel('Wheat Yield (t/ha)')
plt.grid(True)
plt.tight_layout()
plt.show()

```
