## MERGE FIRST

```python
sm.rename(columns={"Marketing Year": "season"},inplace=True)
yields=sp.read_pd_from_excel('sites/grainsargprojects/models/soil_moisture_wheat/yields.xlsx')

# Reformat season column from '2002/03' to '2002 / 2003'
def reformat_season(season_str):
    start_year, end_suffix = season_str.split('/')
    end_year = str(int(start_year[:2] + end_suffix))  # e.g., '2002'[:2] + '03' = '2003'
    return f"{start_year} / {end_year}"

yields['season'] = yields['season'].apply(reformat_season)
```

```python

# Étape 1 : Création de la colonne 'departement' avec mapping
sm['departement'] = sm['name'].apply(lambda x: mapping_dict.get(x, x))

# Merge the DataFrames on the 'departement' column
merged_df = pd.merge(sm, yields, on=['departement','season'], how='left')


province_position = {
    # Northern provinces (0)
    'JUJUY': 0, 'SALTA': 0, 'FORMOSA': 0, 'CHACO': 0, 'SANTIAGO DEL ESTERO': 0, 'MISIONES': 0, 'CORRIENTES': 0,
    
    # Mid-north (1)
    'TUCUMAN': 1, 'CATAMARCA': 1, 'LA RIOJA': 1, 'SAN JUAN': 1, 'ENTRE RIOS': 1, 'SANTA FE': 1,

    # Mid-south (2)
    'CORDOBA': 2, 'BUENOS AIRES': 2, 'LA PAMPA': 2, 'SAN LUIS': 2, 'MENDOZA': 2,

    # Southern provinces (3)
    'RIO NEGRO': 3, 'CHUBUT': 3, 'SANTA CRUZ': 3, 'TIERRA DEL FUEGO': 3,
}

# Convert to uppercase to match mapping keys
merged_df['province'] = merged_df['province'].str.upper()

# Map the provinces to latitude bands
merged_df['lat_band'] = merged_df['province'].map(province_position)
# Ensure date column is datetime
merged_df['date'] = pd.to_datetime(merged_df['date'])

# Define 15-day time windows by lat_band
latband_to_dates = {
    0: ("05-1", "08-1"),  # May 15 – July 15
    1: ("05-15", "08-15"),
    2: ("06-01", "09-1"),  # June 1 – August 15
    3: ("06-15", "09-15")
}

# Extract year from date for building the window
merged_df['year'] = merged_df['date'].dt.year

# Filter based on user parameters
merged_df = merged_df[merged_df['province'].isin(selected_provinces)]
merged_df = merged_df[merged_df['year'] >= min_year]

# Create masks for each lat_band using corresponding 15-day ranges
masks = []
for lat_band, (start_str, end_str) in latband_to_dates.items():
    # Build full date range per row's year
    mask = (
        (merged_df['lat_band'] == lat_band) &
        (merged_df['date'] >= pd.to_datetime(merged_df['year'].astype(str) + "-" + start_str)) &
        (merged_df['date'] <= pd.to_datetime(merged_df['year'].astype(str) + "-" + end_str))
    )
    masks.append(mask)

# Combine all masks
combined_mask = masks[0]
for mask in masks[1:]:
    combined_mask |= mask

# Apply filter
df_filtered = merged_df[combined_mask]

# Assign time window label
latband_to_window = {
    0: "May 1 – Aug 1",
    1: "May 15 – Aug 15",
    2: "June 1 – Sep 1",
    3: "June 15 – Sep 15"
}
df_filtered['time_window'] = df_filtered['lat_band'].map(latband_to_window)

# Group and merge as before
avg_soil_moisture = (
    df_filtered
    .groupby(['province', 'departement', 'season', 'time_window'])['value']
    .mean()
    .reset_index()
    .rename(columns={'value': 'avg_soil_moisture'})
)

avg_soil_moisture_with_yield = avg_soil_moisture.merge(
    yields,
    on=['province', 'departement', 'season'],
    how='left'
)

```

```python
from scipy.stats import zscore
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Copy for safety
data = avg_soil_moisture_with_yield.copy()

# Compute z-scores per departement
data['z_soil'] = data.groupby(['province', 'departement'])['avg_soil_moisture'].transform(lambda x: zscore(x, nan_policy='omit'))
data['z_yield'] = data.groupby(['province', 'departement'])['yield'].transform(lambda x: zscore(x, nan_policy='omit'))

# Flag outliers: z-score > 2 on either axis
data['is_outlier'] = (data['z_soil'].abs() > 2.0) | (data['z_yield'].abs() > 2.0)

# Compute correlation excluding outliers
correlations_cleaned = (
    data[~data['is_outlier']]
    .groupby(['province', 'departement'])
    .apply(lambda x: pd.Series({
        'correlation': x['avg_soil_moisture'].corr(x['yield']),
        'n_years': x['season'].nunique()
    }))
    .reset_index()
)

# Merge back correlation data
data = data.merge(correlations_cleaned, on=['province', 'departement'], how='left')

# Create label

data['prov_dept'] = data['province'] + " - " + data['departement']

# Plotting
sns.set(style="whitegrid")
g = sns.FacetGrid(data, col="prov_dept", col_wrap=4, height=4, sharex=False, sharey=False)

def plot_with_outliers(data, **kwargs):
    ax = plt.gca()
    
    # Plot non-outliers
    sns.scatterplot(data=data[~data['is_outlier']], x="avg_soil_moisture", y="yield", ax=ax, color="blue")
    
    # Plot outliers in red
    outliers = data[data['is_outlier']]
    sns.scatterplot(data=outliers, x="avg_soil_moisture", y="yield", ax=ax, color="red", marker="X", s=80, label="Outlier")
    
    # Annotate outliers with season
    for _, row in outliers.iterrows():
        ax.text(row['avg_soil_moisture'], row['yield'], row['season'], fontsize=8, color="red", ha="left", va="bottom")
    
    # Annotate correlation
    corr = data['correlation'].iloc[0]
    ax.text(0.05, 0.95, f"r = {corr:.2f}", transform=ax.transAxes,
            ha='left', va='top', fontsize=10,
            bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.6))

# Apply to FacetGrid
g.map_dataframe(plot_with_outliers)
g.set_axis_labels("Average Soil Moisture Value", "Yield (kg/ha)")
g.set_titles(col_template="{col_name}")
g.fig.suptitle("Yield vs Soil Moisture per Departement (Outliers Highlighted)", fontsize=16)
plt.subplots_adjust(top=0.93)
plt.show()

# Show sorted correlation table
correlations_cleaned = correlations_cleaned.sort_values(by="correlation", ascending=False)
print(correlations_cleaned)
```

```python
sm.rename(columns={"Marketing Year": "season"},inplace=True)
sm_pro=sm[sm['level']==1]

# Étape 1 : Création de la colonne 'departement' avec mapping
sm_pro['province'] = sm_pro['name'].apply(lambda x: province_map.get(x, x))
sm_pro['name'] = sm_pro['name'].apply(lambda x: province_map.get(x, x))
```

```python
yields_pro=sp.read_pd_from_excel('sites/grainsargprojects/models/soil_moisture_wheat/yield_prov.xlsx')
yields_pro['season'] = yields_pro['season'].apply(reformat_season)
```

```python

# Convert to uppercase to match mapping keys
merged_df['name'] = merged_df['name'].str.upper()

# Map the names to latitude bands
merged_df['lat_band'] = merged_df['name'].map(province_position)
# Ensure date column is datetime
merged_df['date'] = pd.to_datetime(merged_df['date'])

# Define 15-day time windows by lat_band
latband_to_dates = {
    0: ("05-1", "08-1"),  # May 15 – July 15
    1: ("05-15", "08-15"),
    2: ("06-01", "09-1"),  # June 1 – August 15
    3: ("06-15", "09-15")
}


# Filter based on user parameters
merged_df = merged_df[merged_df['province'].isin(selected_provinces)]
merged_df = merged_df[merged_df['year'] >= min_year]


# Extract year from date for building the window
merged_df['year'] = merged_df['date'].dt.year

# Create masks for each lat_band using corresponding 15-day ranges
masks = []
for lat_band, (start_str, end_str) in latband_to_dates.items():
    # Build full date range per row's year
    mask = (
        (merged_df['lat_band'] == lat_band) &
        (merged_df['date'] >= pd.to_datetime(merged_df['year'].astype(str) + "-" + start_str)) &
        (merged_df['date'] <= pd.to_datetime(merged_df['year'].astype(str) + "-" + end_str))
    )
    masks.append(mask)

# Combine all masks
combined_mask = masks[0]
for mask in masks[1:]:
    combined_mask |= mask

# Apply filter
df_filtered = merged_df[combined_mask]

# Assign time window label
latband_to_window = {
    0: "May 1 – Aug 1",
    1: "May 15 – Aug 15",
    2: "June 1 – Aug 1",
    3: "June 15 – Aug 15"
}
df_filtered['time_window'] = df_filtered['lat_band'].map(latband_to_window)

# Group and merge as before
avg_soil_moisture = (
    df_filtered
    .groupby(['province', 'season', 'time_window'])['value']
    .mean()
    .reset_index()
    .rename(columns={'value': 'avg_soil_moisture'})
)

avg_soil_moisture_with_yield = avg_soil_moisture.merge(
    yields_pro,
    on=['province', 'season'],
    how='left'
)


# Set Seaborn style
sns.set(style="whitegrid")

# Create a unique label for each province-departement pair
avg_soil_moisture_with_yield['prov_dept'] = avg_soil_moisture_with_yield['province']
# # OPTIONAL: Limit to top 16 most common prov_dept pairs
# top_prov_dept = avg_soil_moisture_with_yield['prov_dept'].value_counts().nlargest(16).index
# avg_soil_moisture_with_yield_subset = avg_soil_moisture_with_yield[avg_soil_moisture_with_yield['prov_dept'].isin(top_prov_dept)]

avg_soil_moisture_with_yield_subset=avg_soil_moisture_with_yield.copy()

correlations = (
    avg_soil_moisture_with_yield_subset
    .groupby(['province'])
    .apply(lambda x: pd.Series({
        'correlation': x['avg_soil_moisture'].corr(x['yield']),
        'n_years': x['season'].nunique()  # assumes one row per season
    }))
    .reset_index()
)

# Merge back and update label
avg_soil_moisture_with_yield_subset = avg_soil_moisture_with_yield_subset.merge(correlations, on=['province'], how='left')
avg_soil_moisture_with_yield_subset['prov_dept'] = avg_soil_moisture_with_yield_subset['province']

# Create FacetGrid with combined label
g = sns.FacetGrid(avg_soil_moisture_with_yield_subset, col="prov_dept", col_wrap=4, height=4, sharex=False, sharey=False)
g.map_dataframe(sns.scatterplot, x="avg_soil_moisture", y="yield")

# Annotate each subplot with correlation value
def annotate(data, **kwargs):
    corr = data['correlation'].iloc[0]
    ax = plt.gca()
    ax.text(0.05, 0.95, f"r = {corr:.2f}", transform=ax.transAxes,
            ha='left', va='top', fontsize=10, bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.6))

g.map_dataframe(annotate)

# Set axis labels and main title
g.set_axis_labels("Average Soil Moisture Value", "Yield (kg/ha)")
g.set_titles(col_template="{col_name}")
g.fig.suptitle("Yield vs Soil Moisture per Province - Departement", fontsize=16)
plt.subplots_adjust(top=0.93)

plt.show()

# Display correlation table
correlations = correlations.sort_values(by="correlation", ascending=False)
print(correlations)

```

```python

from scipy.stats import zscore

# Recompute province_season_avg with z-score columns
province_season_avg = (
    avg_soil_moisture_with_yield
    .groupby(['province', 'season'], as_index=False)
    .agg({'avg_soil_moisture': 'mean', 'yield': 'mean'})
)

# Add z-score and outlier detection
province_season_avg['z_soil'] = province_season_avg.groupby('province')['avg_soil_moisture'].transform(lambda x: zscore(x, nan_policy='omit'))
province_season_avg['z_yield'] = province_season_avg.groupby('province')['yield'].transform(lambda x: zscore(x, nan_policy='omit'))
province_season_avg['is_outlier'] = (province_season_avg['z_soil'].abs() > 2.0) | (province_season_avg['z_yield'].abs() > 2.0)

# Recalculate cleaned correlations
province_correlations_cleaned = (
    province_season_avg[~province_season_avg['is_outlier']]
    .groupby('province')
    .apply(lambda x: pd.Series({
        'correlation': x['avg_soil_moisture'].corr(x['yield']),
        'n_years': x['season'].nunique()
    }))
)

# Merge correlation back
province_season_avg = province_season_avg.merge(province_correlations_cleaned, on='province', how='left')

# Plot with outliers in red and labeled
g = sns.FacetGrid(province_season_avg, col="province", col_wrap=4, height=4, sharex=False, sharey=False)

def plot_with_outliers(data, **kwargs):
    ax = plt.gca()
    
    # Plot normal points
    sns.scatterplot(data=data[~data['is_outlier']], x="avg_soil_moisture", y="yield", ax=ax, color="blue")
    
    # Plot outliers in red
    outliers = data[data['is_outlier']]
    sns.scatterplot(data=outliers, x="avg_soil_moisture", y="yield", ax=ax, color="red", marker="X", s=80, label="Outlier")
    
    # Annotate outliers
    for _, row in outliers.iterrows():
        ax.text(row['avg_soil_moisture'], row['yield'], row['season'], fontsize=8, color="red", ha="left", va="bottom")
    
    # Add correlation text
    corr = data['correlation'].iloc[0]
    ax.text(0.05, 0.95, f"r = {corr:.2f}", transform=ax.transAxes,
            ha='left', va='top', fontsize=10,
            bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.6))

g.map_dataframe(plot_with_outliers)

g.set_axis_labels("Average Soil Moisture Value", "Yield (kg/ha)")
g.set_titles(col_template="{col_name}")
g.fig.suptitle("Yield vs Soil Moisture per Province (Outliers Highlighted)", fontsize=16)
plt.subplots_adjust(top=0.93)
plt.show()

# Show correlation table
province_correlations_cleaned = province_correlations_cleaned.sort_values(by="correlation", ascending=False)
print(province_correlations_cleaned)
```
