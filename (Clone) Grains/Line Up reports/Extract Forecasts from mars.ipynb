```python
#### Sql keys to codes

country_id = db.sql_read('MarsGrainsReplica',table = 'dbo.Country')
commodity_id = db.sql_read('MarsGrainsReplica', table = 'dbo.Commodity')
commodity_quality_id = db.sql_read('MarsGrainsReplica', table = 'dbo.CommodityQuality')
commodity_sub_quality_id = db.sql_read('MarsGrainsReplica', table = 'dbo.CommoditySubQuality')
status_id = db.sql_read('MarsGrainsReplica', table = 'dbo.Status')
```

```python
#### Rename key dfs to match the bs read in

# Country
country_id["Country"] = country_id["Name"]
country_id["country_id"] = country_id["Id"]

# Commodity
commodity_id["Commodity"] = commodity_id["Name"]
commodity_id["commodity_id"] = commodity_id["Id"]

# Quality
commodity_quality_id["CommodityQuality"] = commodity_quality_id["Name"]
commodity_quality_id["quality_id"] = commodity_quality_id["Id"]
commodity_quality_id["commodity_id"] = commodity_quality_id["Commodity"]

# Subquality
commodity_sub_quality_id["CommoditySubQuality"] = commodity_sub_quality_id["Name"]
commodity_sub_quality_id["subquality_id"] = commodity_sub_quality_id["Id"]
commodity_sub_quality_id["quality_id"] = commodity_sub_quality_id["CommodityQuality"]

# Get rid of extra columns to prevent column name problems
country_id = country_id[["country_id", "Country"]]
commodity_id = commodity_id[["Commodity", "commodity_id"]]
commodity_quality_id = commodity_quality_id[["commodity_id", "quality_id", "CommodityQuality"]]
commodity_sub_quality_id = commodity_sub_quality_id[["CommoditySubQuality", "subquality_id", "quality_id"]]
```

## CORN

```python



# Get rid of extras
tradeflow_corn = tradeflow.loc[tradeflow["Commodity"] == "Corn"]
tradeflow_corn = tradeflow_corn.loc[tradeflow_corn["Quality"] == "All Corn"]
tradeflow_corn = tradeflow_corn.loc[tradeflow_corn["SubQuality"] == "Yellow Corn"]


# # Get rid of null entries
tradeflow_corn = tradeflow_corn.loc[tradeflow_corn["country_id"].notna()]

# # Keep only countries on my list
tradeflow_corn = tradeflow_corn.loc[tradeflow_corn["SourceCountry"].isin(want_countries)]

#### Map status

# Rename status matrix
status_id = status_id.rename({"Id": "status_id",
                  "Name": "Status"}, 
                 axis = 1)

# Merge
tradeflow_corn = pd.merge_ordered(tradeflow_corn, status_id, left_on = ["status_id"], right_on = ["status_id"])

#### Cleanup


# Keep only shipped and at-price
# tradeflow_corn = tradeflow_corn.loc[tradeflow_corn["Status"].isin(["Shipped", "Expected"])]


# Helper column for market year
tradeflow_corn["my_start"] = tradeflow_corn["SourceCountry"].map(cropyearstart_corn)
tradeflow_corn["newcrop"] = tradeflow_corn["Month"] >= tradeflow_corn["my_start"]

# Get market year
tradeflow_corn["my"] = tradeflow_corn["Year"]
tradeflow_corn["my"].loc[~tradeflow_corn["newcrop"]] = tradeflow_corn["my"] - 1



# Get rid of countries without a proper market year mapped
# tradeflow = tradeflow.loc[tradeflow["my_start"].notna()]





```

```python
# Select columns
tradeflow_corn = tradeflow_corn[["SourceCountry",  "DestinationCountry", "Year", "Month", "Value", "my", "newcrop", "my_start", "Status", "IsExporter"]]

def assign_season_corn(row):
    year = row['Year']
    month = row['Month']

    if pd.isna(year) or pd.isna(month):
        return None  # or "Unknown"

    year = int(year)
    month = int(month)

    if month >= 3:
        season_start = year
    else:
        season_start = year - 1

    return f"{season_start}/{season_start + 1}"

# Apply to create the new column
tradeflow_corn['Season'] = tradeflow_corn.apply(assign_season_corn, axis=1)

today = pd.Timestamp.today()

# Determine current season
if today.month >= 3:
    current_season = f"{today.year}/{today.year + 1}"
else:
    current_season = f"{today.year - 1}/{today.year}"

tradeflow_corn_current_season=tradeflow_corn.loc[tradeflow_corn['Season'] == current_season]
tradeflow_corn_current_season=tradeflow_corn_current_season.drop(columns=['SourceCountry','my','my_start','Season','newcrop','IsExporter','Year'])
tradeflow_corn_current_season["DestinationCountry"] = tradeflow_corn_current_season["DestinationCountry"].replace('ZZZ_Unknown_Destination', 'UNKNOWN')
filtered_tradeflow_corn_current_season = tradeflow_corn_current_season[tradeflow_corn_current_season['Status'].isin(['At-price', 'Shipped'])]

filtered_tradeflow_corn_current_season.drop(columns=['Status'])



tradeflow_corn_current_season_pivot = filtered_tradeflow_corn_current_season.pivot_table(
    index='DestinationCountry',
    columns='Month',
    values='Value',
    aggfunc='sum'  # or 'mean', 'first', etc. depending on your need
)

tradeflow_corn_current_season_pivot.columns.name = None
tradeflow_corn_current_season_pivot.columns = tradeflow_corn_current_season_pivot.columns.astype(int)

# Reorder columns: March to December, then January and February
ordered_cols = list(range(3, 13)) + [1, 2]
tradeflow_corn_current_season_pivot = tradeflow_corn_current_season_pivot[ordered_cols]

world_row = tradeflow_corn_current_season_pivot.sum(numeric_only=True)

# Assign it to a new row named 'WORLD'
tradeflow_corn_current_season_pivot.loc['WORLD'] = world_row



tradeflow_corn_current_season_pivot.reset_index(inplace=True)
tradeflow_corn_current_season_pivot['DestinationCountry'] = tradeflow_corn_current_season_pivot['DestinationCountry'].map(country_mapping)
tradeflow_corn_current_season_pivot.rename(columns={'DestinationCountry': '0'}, inplace=True)

```
