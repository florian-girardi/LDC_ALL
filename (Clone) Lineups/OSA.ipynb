```python
from datetime import datetime
import pandas as pd

# Read Excel
osa = sp_mgr.read_pd_from_excel(
    '/sites/grp-oilseedssnd/Shared%20Documents/MainFile/Oilseeds_Analytics_version2.xlsm',
    sheet_name="mtx_sbm"
)

# Keep only columns containing 'arg'
ARG_SBM = osa.loc[:, (osa.columns[0],) + tuple(
    osa.columns[1:][osa.iloc[1, 1:].astype(str).str.contains('arg', case=False, na=False)]
)]

# Set first row as header
ARG_SBM.columns = ARG_SBM.iloc[0]

# Drop the first row
ARG_SBM = ARG_SBM[1:].reset_index(drop=True)

# Drop the next 2 rows (you were doing ARG_SBM = ARG_SBM[2:])
ARG_SBM = ARG_SBM[2:].reset_index(drop=True)

# Now, the first column has the date info.
# Let's get the column name explicitly
date_col_name = ARG_SBM.columns[0]

# Convert to datetime
ARG_SBM[date_col_name] = pd.to_datetime(ARG_SBM[date_col_name], errors='coerce')

# Filter valid datetimes
ARG_SBM = ARG_SBM[ARG_SBM[date_col_name].notna()].reset_index(drop=True)

# Rename the datetime column to 'date'
ARG_SBM = ARG_SBM.rename(columns={date_col_name: 'date'})

# Confirm it's datetime:
assert pd.api.types.is_datetime64_any_dtype(ARG_SBM['date']), "Date column is not datetime"

# Extract month/year
ARG_SBM['month'] = ARG_SBM['date'].dt.month
ARG_SBM['year'] = ARG_SBM['date'].dt.year

# Filter current year
current_year = datetime.now().year
df_current_year = ARG_SBM[ARG_SBM['year'] == current_year]

# Pivot: countries as rows, months as columns
pivot_df = df_current_year.set_index('date').drop(columns=['month', 'year']).T

# Set column names to months
pivot_df.columns = df_current_year['month'].values

# Drop rows with all zeros
pivot_df = pivot_df[~(pivot_df == 0).all(axis=1)]

# Reset index
pivot_df = pivot_df.reset_index().rename(columns={'index': 'Country'})

pivot_df
```
